
# PCA — приклад саме з **цими** вагами (loadings)

Мета:
- Показати ваги рівно як на слайді
- Показати, що **PC1** легко інтерпретувати (є явні топ-фічі)
- Показати, що **PC2** інтерпретувати важче (менш “чистий” сенс)
- Пояснити, що `head(k)` — це просто скільки топ-фіч показувати


```

import numpy as np
import pandas as pd
```


## 1) Фіксовані ваги (як на твоєму прикладі)
PCA в реальності повертає `PC1`, `PC2` і їхні ваги. Тут ми **зафіксуємо** ваги вручну, щоб вони були точнісінько такі ж.


```

loadings = pd.DataFrame(
    [
        [ 0.55,  0.50,  0.48,  0.20, -0.40],  # PC1
        [-0.10,  0.15,  0.20,  0.70,  0.65],  # PC2
    ],
    columns=["size", "rooms", "bathrooms", "schools", "crime_rate"],
    index=["PC1", "PC2"]
)

loadings
```


## 2) Як знайти “найважливіші” фічі
Ми дивимось на **абсолютні** значення (бо знак = напрям, а сила = модуль).


```

abs_loadings = loadings.abs()
abs_loadings
```


### PC1 — “видно” найважливіші
Для PC1 у нас одразу кидаються в очі 3 найбільші коефіцієнти: `size`, `rooms`, `bathrooms`.


```

# Покажемо ТОП-3 для PC1
abs_loadings.loc["PC1"].sort_values(ascending=False).head(3)
```


### PC2 — “не видно” так само чітко (інтерпретація менш очевидна)
У PC2 найбільші `schools` і `crime_rate`, але сенс може бути “змішаний” (соціально-локаційний фактор),
і назва вже не така очевидна, як для PC1.


```

# Покажемо ТОП-2 для PC2 (щоб було видно, що домінують schools/crime_rate)
abs_loadings.loc["PC2"].sort_values(ascending=False).head(2)
```


## 3) Важливо: `head(k)` — це не PCA, це *як багато* ти хочеш показати
Якщо зробиш `head(2)` — побачиш 2 фічі. Якщо `head(3)` — 3 фічі і т.д.


```

for k in [2, 3, 5]:
    print(f"\nTOP-{k} for PC1:")
    print(abs_loadings.loc["PC1"].sort_values(ascending=False).head(k))
```


## 4) Як порахувати значення PC1 для конкретного будинку
Це просто лінійна комбінація (проєкція):

PC1_value = 0.55*size + 0.50*rooms + 0.48*bathrooms + 0.20*schools - 0.40*crime_rate

Нижче — приклад для одного “будинку” (одного рядка).


```

# Один приклад будинку (умовні значення фіч)
x = pd.Series({"size": 120, "rooms": 4, "bathrooms": 2, "schools": 6, "crime_rate": 40})
x
```

```

pc1_value = (loadings.loc["PC1"] * x).sum()
pc2_value = (loadings.loc["PC2"] * x).sum()

pc1_value, pc2_value
```


### Чому ми “перемножаємо”?
Бо це **скалярний добуток**: точка (фічі будинку) · напрям (PC1).  
Так ми отримуємо координату точки вздовж нової осі PC1/PC2.
