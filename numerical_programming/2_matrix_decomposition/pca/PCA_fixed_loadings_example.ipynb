
# PCA Example (With Fixed Loadings)

This notebook uses the exact loadings shown in the example:

PC1 = [0.55, 0.50, 0.48, 0.20, -0.40]  
PC2 = [-0.10, 0.15, 0.20, 0.70, 0.65]

We will:
- Start with 5D data
- Manually define PCA components
- Project data to 2D
- Verify projection using matrix multiplication


```

import numpy as np
import pandas as pd
```

```

np.random.seed(0)

data = pd.DataFrame({
    "size": np.random.normal(100, 20, 5),
    "rooms": np.random.normal(3, 0.5, 5),
    "bathrooms": np.random.normal(2, 0.3, 5),
    "schools": np.random.normal(5, 2, 5),
    "crime_rate": np.random.normal(50, 10, 5)
})

data
```

```

X = (data - data.mean()) / data.std()
X
```

```

components = np.array([
    [0.55, 0.50, 0.48, 0.20, -0.40],
    [-0.10, 0.15, 0.20, 0.70, 0.65]
])

loadings = pd.DataFrame(
    components,
    columns=data.columns,
    index=["PC1", "PC2"]
)

loadings
```


## Project 5D → 2D

New coordinates are computed as:

X_2D = X_scaled @ components.T


```

X_2d = X.values @ components.T
df_pca = pd.DataFrame(X_2d, columns=["PC1", "PC2"])
df_pca
```


## Interpretation

PC1 is mostly influenced by:
- size
- rooms
- bathrooms

PC2 is mostly influenced by:
- schools
- crime_rate

Weights show how strongly each original feature contributes 
to the new principal component.
